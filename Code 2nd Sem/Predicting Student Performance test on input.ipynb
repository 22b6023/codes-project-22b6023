```python
import pandas as pd
from pycaret import classification as pyclf


MODEL_PATH = r"C:\Users\User\Desktop\Final Year Project\Data\Best_RiskClassifier_EarlyWeighted"
model = pyclf.load_model(MODEL_PATH)

# Prompt 
print("Enter student's GPA (0.00 - 5.00 scale):")
regno = input("Student REGNO (optional): ") or "N/A"

GPA_Y1S1 = float(input("GPA Year 1 Sem 1: "))
GPA_Y1S2 = float(input("GPA Year 1 Sem 2: "))
GPA_Y2S1 = float(input("GPA Year 2 Sem 1: "))
GPA_Y2S2 = float(input("GPA Year 2 Sem 2: "))

# Calculate CGPA summaries 
Y1_avg = round((GPA_Y1S1 + GPA_Y1S2) / 2, 2)
Y2_avg = round((GPA_Y2S1 + GPA_Y2S2) / 2, 2)
overall_avg = round((GPA_Y1S1 + GPA_Y1S2 + GPA_Y2S1 + GPA_Y2S2) / 4, 2)
trend = round(Y2_avg - Y1_avg, 2)

# dataframe for prediction
student_df = pd.DataFrame([{
    'REGNO': regno,
    'GPA_Y1S1': GPA_Y1S1,
    'GPA_Y1S2': GPA_Y1S2,
    'GPA_Y2S1': GPA_Y2S1,
    'GPA_Y2S2': GPA_Y2S2
}])

# Predict
pred = pyclf.predict_model(model, data=student_df).copy()

pred_col = next(c for c in ['prediction_label','Label','prediction'] if c in pred.columns)
risk_label = pred.loc[0, pred_col]


print("\nPrediction Results")
print(f"Student: {regno}")
print(f"Predicted Risk Level: {risk_label}\n")

print("GPA Summary ")
print(f"Year 1 Avg GPA: {Y1_avg}")
print(f"Year 2 Avg GPA: {Y2_avg}")
print(f"Overall Avg GPA (Y1–Y2): {overall_avg}")
print(f"GPA Trend (Y2 - Y1): {trend:+.2f}")

# Simple interpretation based on trend
if trend > 0.1:
    print("Performance improving across semesters.")
elif trend < -0.1:
    print("Performance declining; risk may increase if trend continues.")
else:
    print("Performance stable with minor variation.")
```
