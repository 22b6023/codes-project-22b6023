Dataset 1: Demographics (Personal Details).

```python
from pathlib import Path
import pandas as pd
import numpy as np


PROJECT_DIR = Path(r"C:\Users\User\Desktop\Final Year Project")
DATA_DIR    = PROJECT_DIR / "Data"
OUT_DIR     = PROJECT_DIR / r"22b6023 Muizzul" / "3 Processed Data"
OUT_DIR.mkdir(parents=True, exist_ok=True)


PERSONAL_PATH = DATA_DIR / "ListOfStudPersDetails.xlsx"

print("DATA_DIR:", DATA_DIR)
print("OUT_DIR:", OUT_DIR)
```

```python
df_raw = pd.read_excel(PERSONAL_PATH)
print(df_raw.shape)
df_raw.head()
```

```python
rename_map = {
    "Reg No": "REGNO",
    "Faculty": "Faculty",
    "Degree Type": "DegreeType",
    "Status": "Status",
    "Title": "Title",
    "Gender": "Gender",
    "Degree": "Degree",
    "Date of Birth": "DOB",
    "Citizenship": "Citizenship",
    "Race": "Race",
    "Religion": "Religion",
    "Funding": "Funding",
    "Year Joined": "YearJoined",
    "Part Time": "PartTime",
    "Major code": "MajorCode",
    "Major name": "MajorName",
    "Minor": "Minor",
    "First Registered": "FirstRegistered",
    "Last Registered": "LastRegistered",
}

cols_keep = list(rename_map.keys())
df = df_raw[cols_keep].rename(columns=rename_map)
df.head()
```

```python
# whitespace
for c in df.select_dtypes(include="object").columns:
    df[c] = df[c].astype(str).str.strip()

# REGNO as nullable integer 
df["REGNO"] = pd.to_numeric(df["REGNO"], errors="coerce").astype("Int64")

# Parse date columns 
for dcol in ["DOB", "FirstRegistered", "LastRegistered"]:
    df[dcol] = pd.to_datetime(df[dcol], errors="coerce")

# Standardize Gender to {Male, Female, Unknown}
gender_map = {
    "M": "Male", "F": "Female",
    "male": "Male", "female": "Female",
    "MALE": "Male", "FEMALE": "Female"
}
df["Gender"] = df["Gender"].replace(gender_map)
df.loc[~df["Gender"].isin(["Male", "Female"]), "Gender"] = "Unknown"

# Standardize PartTime to {Yes, No, Unknown}
df["PartTime"] = df["PartTime"].str.upper().map({"YES": "Yes", "NO": "No"})
df.loc[~df["PartTime"].isin(["Yes", "No"]), "PartTime"] = "Unknown"

# Fill low-importance missing with 'Unknown'
for c in ["Race", "Religion", "Funding", "MajorCode", "Minor"]:
    df[c] = df[c].replace({None: "Unknown", "nan": "Unknown"}).fillna("Unknown")

df.head()
```

```python
# Drop rows without a REGNO (since we can't link them to other tables, and they are few)
df = df[~df["REGNO"].isna()].copy()

# Keep the most recent record per student
df.sort_values(["REGNO", "LastRegistered"], ascending=[True, False], inplace=True)
df = df.drop_duplicates(subset=["REGNO"], keep="first").reset_index(drop=True)

print("After dedup:", df.shape)
```

```python
today = pd.Timestamp.today().normalize()

df["Age"] = ((today - df["DOB"]).dt.days / 365.25).round(1)
df["YearsSinceFirstReg"] = ((today - df["FirstRegistered"]).dt.days / 365.25).round(1)
df["YearsSinceLastReg"]  = ((today - df["LastRegistered"]).dt.days / 365.25).round(1)

df[["Age", "YearsSinceFirstReg", "YearsSinceLastReg"]].describe()
```

```python
print("Rows:", len(df))

print("\nNulls per column:")
print(df.isna().sum().sort_values(ascending=False))

print("\nGender distribution:")
print(df["Gender"].value_counts(dropna=False))

print("\nPartTime distribution:")
print(df["PartTime"].value_counts(dropna=False))

df.head(10)
```

```python
out_path = OUT_DIR / "personal_clean.csv"
df.to_csv(out_path, index=False)
print("Saved:", out_path)
```

Transcript Records preprocessing

```python
from pathlib import Path
import pandas as pd
import numpy as np


PROJECT_DIR = Path(r"C:\Users\User\Desktop\Final Year Project")
DATA_DIR    = PROJECT_DIR / "Data"
OUT_DIR     = PROJECT_DIR / r"22b6023 Muizzul" / "3 Processed Data"
OUT_DIR.mkdir(parents=True, exist_ok=True)


TRANSCRIPT_PATH = DATA_DIR / "Transcript.xlsx"

print("Reading:", TRANSCRIPT_PATH)
print("Saving to:", OUT_DIR)
```

```python
df_raw = pd.read_excel(TRANSCRIPT_PATH)
print(df_raw.shape)
df_raw.head()
```

```python
rename_map = {
    "REGNO": "REGNO",
    "MODULE_CODE": "ModuleCode",
    "MODULE_NAME": "ModuleName",
    "PASS_FAIL": "PassFail",
    "MODULE_ACADEMIC_REMARKS": "Remarks",
    "MC": "MC",
    "GPA_CALC": "GPA_calc",
    "CGPA_CALC": "CGPA_calc",
    "MARKS": "Marks",
    "P_C_W": "PracticalCW",
    "C_W": "Coursework",
    "SGPA_CALC": "SGPA_calc",
    "GRADE_NAME": "GradeName",
    "GRADE_PTS": "GradePts",
    "LEVEL_WEIGHT": "LevelWeight"
}

df = df_raw.rename(columns=rename_map)[list(rename_map.values())]
df.head()
```

```python
df["PassFail"] = df["PassFail"].fillna("Unknown").str.strip()
df["GradeName"] = df["GradeName"].fillna("Unknown").str.strip()
df["Remarks"] = df["Remarks"].fillna("None")
```

```python
def map_passfail(x):
    if pd.isna(x) or x == "Unknown":
        return -1
    if str(x).upper() == "PASS":
        return 1
    if str(x).upper() == "FAIL":
        return 0
    return -1  # catch-all

df["PassFailBinary"] = df["PassFail"].apply(map_passfail)
df["PassFailBinary"].value_counts()
```

```python
df_summary = df.groupby("REGNO").agg({
    "ModuleCode": "count",
    "PassFailBinary": lambda x: (x==1).mean(),   # pass rate
    "Marks": "mean",
    "GradePts": "mean"
}).reset_index()

df_summary = df_summary.rename(columns={
    "ModuleCode": "TotalModules",
    "PassFailBinary": "PassRate",
    "Marks": "AvgMarks",
    "GradePts": "AvgGradePts"
})

df_summary.head()
```

```python
out_path_detail = OUT_DIR / "transcript_clean.csv"
out_path_summary = OUT_DIR / "transcript_summary.csv"

df.to_csv(out_path_detail, index=False)
df_summary.to_csv(out_path_summary, index=False)

print("Saved:", out_path_detail)
print("Saved:", out_path_summary)
```
